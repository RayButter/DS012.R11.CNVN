```python
import numpy as np


def f(u):
    x, y = u
    return x**4 + y**2 + 2*x*y + 1


def grad(u):
    x, y = u
    return np.array([4*x**3 + 2*y, 2*y + 2*x], dtype=float)


def hessian(u):
    x, y = u
    return np.array([[12*x**2, 2], [2, 2]], dtype=float)


def newton_optimizer(grad_fn, hessian_fn, u_0, tol=1e-6, max_iter=100):
    u = np.array(u_0, dtype=float)
    history = [u.copy()]

    for _ in range(max_iter):
        g = grad_fn(u)
        if np.linalg.norm(g) < tol:
            break

        H = hessian_fn(u)
        if np.linalg.cond(H) > 1e12:
            print("Hessian gan suy bien, khong the tiep tuc")
            break

        a = np.linalg.solve(H, g)
        u = u - a
        history.append(u.copy())

    return history, u


def phan_loai(u, grad_fn, hessian_fn, tol=1e-6, eig_tol=1e-5):
    g = grad_fn(u)
    if not np.all(np.isfinite(u)) or not np.all(np.isfinite(g)):
        return "Ket qua khong huu han, chua the ket luan"
    if np.linalg.norm(g) >= tol:
        return "Chua hoi tu toi diem dung"

    eig = np.linalg.eigvalsh(hessian_fn(u))
    if np.all(eig > eig_tol):
        return "Cuc tieu"
    if np.all(eig < -eig_tol):
        return "Cuc dai"
    if np.any(eig < -eig_tol) and np.any(eig > eig_tol):
        return "Diem yen ngua"
    return "Chua xac dinh bang tieu chuan Hessian"


def in_ket_qua(u_0):
    history, u = newton_optimizer(grad, hessian, u_0)
    print("Diem bat dau:", u_0)
    print("Diem cuoi:", u)
    print("Gia tri ham:", f(u))
    print("So lan cap nhat:", len(history) - 1)
    print("Do lon gradient:", np.linalg.norm(grad(u)))
    print("Ket luan:", phan_loai(u, grad, hessian))
    return history, u


if __name__ == "__main__":
    np.set_printoptions(precision=10, suppress=True)
    # Thay hai tọa độ này để thử điểm đầu khác.
    u_0 = (10, 10)
    history, u = in_ket_qua(u_0)
```
